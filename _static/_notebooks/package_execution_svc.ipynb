# Sephora TF-IDF + LinearSVC execution


```python
import pandas as pd
from sephora_test.utils.train_test_split import (
    assign_train_validation_split,
)

from sephora_test.apply_transformation.clean_text import clean_text
from sephora_test.run_experiments.run_ml_pipeline import (
    config,
    run_ml_pipeline,
)
```

## Load datasets


```python
training_data = pd.read_csv(config["data_paths"]["training_data"])

testing_data = pd.read_csv(config["data_paths"]["test_data"])

print("Training shape:", training_data.shape)
print("Testing shape:", testing_data.shape)
```

## Text preprocessing


```python
training_data = training_data.copy()
testing_data = testing_data.copy()

training_data["cleanName"] = (
    training_data["productName"].fillna("").apply(clean_text)
)

training_data["cleanDescription"] = (
    training_data["longDescription"].fillna("").apply(clean_text)
)

testing_data["cleanName"] = (
    testing_data["productName"].fillna("").apply(clean_text)
)

testing_data["cleanDescription"] = (
    testing_data["longDescription"].fillna("").apply(clean_text)
)
```

## Train / validation split


```python
training_data = assign_train_validation_split(
    training_data,
    test_size=config["validation_size"],
    random_state=config["random_state"],
)

training_data["sample"].value_counts()
```

## Run TF-IDF + LinearSVC pipeline


```python
results = run_ml_pipeline(
    training_data=training_data,
    testing_data=testing_data,
    experiment_config=config,
)
```
